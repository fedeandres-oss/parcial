# Análisis de Datos y Conexión a PostgreSQL
Este cuaderno verifica la conexión a la base de datos de Joomla y realiza una consulta de métricas de actividad.

```python
import os
import pandas as pd
from sqlalchemy import create_engine

# Parámetros de conexión desde la red interna (backend_net)
DB_USER = os.getenv('POSTGRES_USER', 'joomlauser')
DB_PASS = os.getenv('POSTGRES_PASSWORD', 'joomlapassword')
DB_HOST = 'database'
DB_PORT = '5432'
DB_NAME = os.getenv('POSTGRES_DB', 'joomladb')

# Crear motor de conexión con SQLAlchemy
engine = create_engine(f'postgresql://{DB_USER}:{DB_PASS}@{DB_HOST}:{DB_PORT}/{DB_NAME}')
print("Conexión configurada correctamente hacia PostgreSQL.")
```

```python
# Probar consulta de tablas existentes en la base de datos de Joomla
try:
    query = "SELECT table_name FROM information_schema.tables WHERE table_schema='public';"
    df_tables = pd.read_sql(query, con=engine)
    print(f"Se encontraron {len(df_tables)} tablas en la base de datos:")
    print(df_tables.head(10))
except Exception as e:
    print(f"Error al conectar o consultar la base de datos: {e}")
```
