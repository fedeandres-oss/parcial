# Análisis de Datos y Conexión a PostgreSQL
Este cuaderno verifica la conexión a la base de datos de Joomla y realiza una consulta de métricas de actividad.

```python
import warnings
import pandas as pd
import psycopg2

# Ocultar advertencias de Pandas
warnings.filterwarnings('ignore', category=UserWarning)

# 1. Parámetros de conexión
db_user = 'joomlauser'
db_pass = 'joomlapassword'
db_host = 'database'
db_port = '5432'
db_name = 'joomladb'

# 2. Conexión con psycopg2
conn = psycopg2.connect(
    dbname=db_name,
    user=db_user,
    password=db_pass,
    host=db_host,
    port=db_port
)

# 3. Script para listar SOLO las tablas que tienen datos
cursor = conn.cursor()
cursor.execute("SELECT table_name FROM information_schema.tables WHERE table_schema = 'public';")
tablas = [row[0] for row in cursor.fetchall()]

print("--- TABLAS CON REGISTROS EN JOOMLADB ---")
for tabla in tablas:
    cursor.execute(f'SELECT COUNT(*) FROM "{tabla}"')
    total = cursor.fetchone()[0]
    if total > 0:
        print(f"Tabla: {tabla:<35} | Registros: {total}")

conn.close()
```
